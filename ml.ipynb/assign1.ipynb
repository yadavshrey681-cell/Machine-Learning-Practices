```python
import pandas as pd

df = pd.DataFrame(data)

print(type(df))#for colums name
print(df.columns.tolist())
```

```python
#clasify all columns  name
Identifier = "Student_ID"
Categorical = "Name"
numerical= ["Age","Hours_Studied","Attendance","Marks"]
print('Identifier:',Identifier)
print('Categorical:',Categorical)
print('numerical:',numerical)
```

```python
average= df['Marks'].mean()
result=df[df['Marks']>average]
print(result)
print('average =',average)
```
