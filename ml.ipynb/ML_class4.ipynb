```python
import numpy as np
from sklearn.preprocessing import PowerTransformer

X = np.array([
    [-1],
    [0],
    [3],
    [5],
    [10],
    [50],
])
pt= PowerTransformer(method='yeo-johnson')

X_transformed = pt.fit_transform(X)
print('Original:')
print(X.ravel())

print('\nTransformed:')
print(np.round(X_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer

y = np.array([
    [1],
    [1],
    [3],
    [5],
    [10],
    [50],
])
pt= PowerTransformer(method='box-cox')

y_transformed = pt.fit_transform(y)
print('Original:')
print(y.ravel())

print('\nTransformed:')
print(np.round(y_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```
