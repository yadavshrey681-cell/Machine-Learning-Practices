```python
array= np.linspace(12,22, num=15, dtype= int)
```

```python
np.var(array)
```

```python
np.std(array)
```

```python
def variance (*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_variance= total/(len(args))
    return calculated_variance,calculated_variance**(0.5)
```

```python
variance(*array)
```

```python
def variance_sample (*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_variance= total/(len(args)-1)
    return calculated_variance,calculated_variance**(0.5)
```

```python
variance_sample(*array)
```

```python
def standard_deviation(*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_standard= total/(len(args))
    return calculated_standard,calculated_standard**(0.5)
```

```python
standard_deviation(*array)
```
